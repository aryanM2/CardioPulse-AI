```python
# 1. Start fresh with a clean copy
df2 = df.copy()

# 2. Binary Mappings
df2['Sex'] = df2['Sex'].map({"M": 1, "F": 0})
df2['ExerciseAngina'] = df2['ExerciseAngina'].map({'N': 0, 'Y': 1})

# 3. One-Hot Encoding for multi-category columns
df2 = pd.get_dummies(df2, columns=['ChestPainType', 'RestingECG', 'ST_Slope'], drop_first=True, dtype=int)

# 4. Clean Renaming
df2.rename(columns={'Sex': 'isMale', 'ExerciseAngina': 'isExerciseAngina'}, inplace=True)

# 5. Fix invalid clinical zeros (Optional but highly recommended)
df2['RestingBP'] = df2['RestingBP'].replace(0, df2['RestingBP'].median())
df2['Cholesterol'] = df2['Cholesterol'].replace(0, df2['Cholesterol'].median())

df2.head()
```

```python
df2['Cholesterol'].value_counts()
```

```python
# 1. Calculate the mean for Cholesterol excluding 0 values
ch_mean = df2.loc[df2['Cholesterol'] != 0, 'Cholesterol'].mean()
df2['Cholesterol'] = df2['Cholesterol'].replace(0, ch_mean).round(2)

# 2. Calculate the mean for RestingBP excluding 0 values (Fixing the case typo: 'RestingBP')
resting_bp_mean = df2.loc[df2['RestingBP'] != 0, 'RestingBP'].mean()
df2['RestingBP'] = df2['RestingBP'].replace(0, resting_bp_mean).round(2)

# 3. View the final result
df2.head()
```

```python
from sklearn.preprocessing import StandardScaler
numeric_col = ['Age','RestingBP','Cholesterol','MaxHR','Oldpeak']
scaler = StandardScaler()
df_encoded[numeric_col] = scaler.fit_transform(df_encoded[numeric_col])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, f1_score, confusion_matrix
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
```

```python
X = df_encoded.drop('HeartDisease', axis=1)
y = df_encoded['HeartDisease']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.23, random_state=42)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
models={
    "LogisticRegression":LogisticRegression(),
    "KNeighborsClassifier":KNeighborsClassifier(),
    "GaussianNB":GaussianNB(),
    "DecisionTreeClassifier":DecisionTreeClassifier(),
    "SVC":SVC()
}
```

```python
results = []
```

```python
for name,model in models.items():
  model.fit(X_train_scaled,y_train)
  y_pred = model.predict(X_test_scaled)
  accuracy = accuracy_score(y_test,y_pred)
  f1 = f1_score(y_test,y_pred)
  results.append({
      'model':name,
      'accuracy':round(accuracy,4),
      "f1_score":round(f1,4)
  })
```

```python
results
```

```python
import joblib
joblib.dump(models['LogisticRegression'],'model.pkl')
joblib.dump(scaler,'scaler.pkl')
joblib.dump(X.columns.tolist(),'columns.pkl')
```
